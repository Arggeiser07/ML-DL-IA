```python
# Descargamos el modelo en español (el modelo 'lg' o 'md' tienen vectores, el 'sm' no)
!python -m spacy download es_core_news_md

import spacy
import numpy as np
from sklearn.metrics.pairwise import cosine_similarity

# Cargamos el modelo
nlp = spacy.load("es_core_news_md")
```

```python
#Dimensiones de un vector ( palabra)
palabra = "criptomoneda"
vector = nlp(palabra).vector

print(f"Palabra: {palabra}")
print(f"Dimensiones del vector: {vector.shape}")
print(f"Primeros 50 números del vector:\n{vector[:50]}")
```

```python
#Similitud semántica
def comparar(p1, p2):
    v1 = nlp(p1).vector.reshape(1, -1)
    v2 = nlp(p2).vector.reshape(1, -1)
    similitud = cosine_similarity(v1, v2)[0][0]
    print(f"Similitud entre '{p1}' y '{p2}': {similitud:.4f}")

# Pruebas de cercanía semántica
comparar("perro", "gato")
comparar("perro", "computadora")
comparar("bitcoin", "dinero")
comparar("rey", "reina")
comparar("topo", "retrete")
comparar("boquete", "agujero")
```

```python
# --- EL DESAFÍO DE LA REINA (MÉTODO SEGURO) ---
# Calculamos el vector teórico: Rey - Hombre + Mujer
v_objetivo = nlp("rey").vector - nlp("hombre").vector + nlp("mujer").vector

# Lo comparamos directamente con las opciones lógicas
print("\n--- TEST DE ÁLGEBRA ---")
for opcion in ["reina", "rey", "trono", "mujer"]:
    v_opcion = nlp(opcion).vector.reshape(1, -1)
    score = cosine_similarity(v_objetivo.reshape(1, -1), v_opcion)[0][0]
    print(f"Parecido con '{opcion}': {score:.4f}")

```

```python
# 1. Definimos los vectores necesarios
v_rey = nlp("rey").vector
v_hombre = nlp("hombre").vector
v_mujer = nlp("mujer").vector

# 2. Creamos la variable que te faltaba
v_resultado = v_rey - v_hombre + v_mujer

# 3. Comparamos con palabras nuevas
opciones_nuevas = ["reina", "princesa", "monarca", "presidenta"]

print("--- SEGUNDO INTENTO ---")
for opcion in opciones_nuevas:
    v_opcion = nlp(opcion).vector.reshape(1, -1)
    score = cosine_similarity(v_resultado.reshape(1, -1), v_opcion)[0][0]
    print(f"Parecido con '{opcion}': {score:.4f}")
```

```python
# Analogía de países: Madrid es a España lo que X es a Francia
v_madrid = nlp("madrid").vector
v_espana = nlp("españa").vector
v_francia = nlp("francia").vector

v_ciudad_misteriosa = v_madrid - v_espana + v_francia

# ¿Saldrá París?
score_paris = cosine_similarity(v_ciudad_misteriosa.reshape(1, -1), nlp("parís").vector.reshape(1, -1))[0][0]
print(f"\nSimilitud del resultado con 'París': {score_paris:.4f}")
```

```python
#Buscador semántico

# 1. Tu base de conocimientos
preguntas_db = [
    "¿Cómo puedo recuperar mi contraseña?",
    "¿Dónde están mis facturas?",
    "¿Cuál es el horario de atención al cliente?",
    "Quiero cancelar mi suscripción"
]

# 2. Lo que el usuario pregunta (con otras palabras)
usuario = "No recuerdo mi clave de acceso"

# 3. Comparación inteligente
print(f"Buscando ayuda para: '{usuario}'\n")

for pregunta in preguntas_db:
    similitud = nlp(usuario).similarity(nlp(pregunta))
    print(f"-> Coincidencia con '{pregunta}': {similitud:.4f}")
```
