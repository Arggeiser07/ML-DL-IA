Proceso detallado desde la entrada de un texto intacto ( correo, usuario ) hasta su salido, siendo validado como spam/No spam.
Proyecto de aprendizaje con todas las fases por las que pasa un texto antes de ser "comprendido" por una máquina.

```python
#Importamos herramientas y BD lingüísticas.
import nltk
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
import re

# Descargamos los paquetes de datos necesarios
nltk.download('punkt')
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('omw-1.4')
nltk.download('punkt_tab')


print("✅ Cirujanos listos para operar")
```

```python
# Nuestro espécimen sucio
texto_sucio = '''¡¡¡HOLA!!! 123 Este es un EJEMPLO de texto... ¿acaso no es GENIAL?
Visita https://nlp.com y aprende @nlp #IA 2024'''

print(f"Texto original: {texto_sucio}")
```

```python
#Empezamos a limpiar...
# 1. Todo a minúsculas (para que 'HOLA' y 'hola' sean lo mismo)
texto_limpio = texto_sucio.lower()

# 2. Quitar números y signos de puntuación usando 're' (Expresiones Regulares)
# Este comando dice: "Busca todo lo que NO sea una letra (a-z) y sustitúyelo por un espacio"
texto_limpio = re.sub(r'[^a-z\sáéíóúñ]', '', texto_limpio)

print(f"Texto tras cirugía: {texto_limpio}")
```

```python
#Importamos tokenizer
from nltk.tokenize import word_tokenize

# Usamos la función de NLTK para trocear el texto. No se usa texto.split de python
tokens = word_tokenize(texto_limpio)

print(f"Número de tokens: {len(tokens)}")
print(f"Lista de palabras: {tokens}")
```

```python
# Traemos la lista de palabras prohibidas en español
stop_words = set(stopwords.words('spanish'))

# Filtramos: "Solo quédate con la palabra si NO está en la lista de stop_words"
tokens_interesantes = [word for word in tokens if word not in stop_words]

print(f"Tokens originales: {tokens}")
print(f"Tokens tras limpieza: {tokens_interesantes}")
```

```python
#Lemantización
# Inicializamos al "reductor"
lemmatizer = WordNetLemmatizer()

# Aplicamos la lematización a cada palabra de nuestra lista limpia
tokens_raíz = [lemmatizer.lemmatize(word) for word in tokens_interesantes]

print(f"Tokens limpios: {tokens_interesantes}")
print(f"Tokens en su raíz: {tokens_raíz}")
```

```python
#Creando BoW
from sklearn.feature_extraction.text import CountVectorizer

# 1. Preparamos unas frases de ejemplo (o usamos la nuestra)
corpus = [
    "la ia es genial",
    "el nlp es genial",
    "ia y nlp son el futuro",
    "hola  este es un ejemplo de texto acaso no es genial"
]

# 2. Creamos el vectorizador
vectorizador = CountVectorizer()

# 3. Transformamos el texto en números
X = vectorizador.fit_transform(corpus)

# 4. Ver qué palabras ha aprendido
print("Vocabulario:", vectorizador.get_feature_names_out())

# 5. Ver la matriz resultante
print("Matriz:\n", X.toarray())
```

```python
#Ya hemos probado nuestro vectorizador ahora hagamos una prubea para detectar spam.

# 1. Creamos un mini-dataset
mensajes = [
    "¡Gana un premio gratis ahora!",        # Spam
    "Reunión de trabajo a las diez",        # Ham
    "Haz clic para recibir tu premio",      # Spam
    "¿Quieres ir a comer mañana?",          # Ham
    "Oferta exclusiva: dinero gratis hoy"    # Spam
]

# Etiquetas: 1 para Spam, 0 para Ham
etiquetas = [1, 0, 1, 0, 1]

# 2. Convertimos los mensajes a números (como hiciste antes)
X = vectorizador.fit_transform(mensajes)

print("Vocabulario de entrenamiento:", vectorizador.get_feature_names_out())
```

```python
# 1. El Cerebro (Naive Bayes)
from sklearn.naive_bayes import MultinomialNB
modelo = MultinomialNB()
modelo.fit(X, etiquetas) # Entrenamos con los datos anteriores

# 2. Tu turno: Escribe aquí abajo lo que quieras probar
# Prueba con algo como: "Felicidades, ganaste dinero hoy"
# O con algo como: "Mañana hay reunión de equipo"
mi_test = ["Regalo cripto"]

# 3. La IA lo analiza
# IMPORTANTE: Usamos .transform para usar el mismo "mapa" que aprendió antes
test_X = vectorizador.transform(mi_test)
resultado = modelo.predict(test_X)

# 4. El veredicto
print(f"Mensaje analizado: '{mi_test[0]}'")
if resultado[0] == 1:
    print("🚨 RESULTADO: ¡SPAM DETECTADO! (La IA no se fía)")
else:
    print("✅ RESULTADO: Mensaje limpio. (Pasa el filtro)")
```
